```python
# from ast import Continue
# Task = 3

raw_data = ["  Delhi", "mumbai ", "PUNE", "  delhi ", "Mumbai"]

clean_data = []

for city in raw_data:
  if city.strip().lower() in clean_data:
    continue
  else:
    clean_data.append(city.strip().lower())

print(clean_data)
```
