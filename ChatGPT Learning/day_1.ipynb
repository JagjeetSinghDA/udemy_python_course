```python
numbers = [1,2,3,4,5]

even_numbers_squares = []
for n in numbers:
  if n % 2 == 0:
    even_numbers_squares.append(n * n)

print(even_numbers_squares)
```

```python
nums = [10, 15, 20, 25, 30]
even_squares = []

for n in nums:
  if (n * n) % 2 == 0:
    even_squares.append(n * n)

print(even_squares)
```

```python
items = ["apple", "banana", "apple", "orange", "banana", "apple"]

freq = {}
for item in items:
  if item in freq:
    freq[item] = freq[item] + 1 
  else:
    freq[item] = 1 
  
print(freq)
```

```python
scores = [35, 78, 90, 45, 60, 88]

new_list_over_60 = []

for score in scores:
  if score > 60:
    new_list_over_60.append(score)

print(new_list_over_60)
```
