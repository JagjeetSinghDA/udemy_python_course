```python
t = (1, 2, 3)
t[0] = 100

# tuple is immutable so we can not assign any value at any place or make any changes in any element
```

```python
nums = [5, 10, 15]

tuple_nums = tuple(nums)
print(nums)
print(tuple_nums)
```

```python
nums = [1, 2, 2, 3, 4, 4, 5]

set_nums = set(nums)
print(nums)
print(set_nums)
```

```python
cities = {"delhi", "mumbai", "pune"}

print('delhi' in cities)
print('chennai' in cities)
```

```python
d1 = {
    "name": 'Ginny'
    , "age": 28
    , "city": 'Ghaziabad'
}

print(d1)
```

```python
print(d1["name"])

d1["profession"] = 'Private employee'
print(d1)

d1["age"] = d1["age"] + 1

print(d1)
```

```python
items = ["a", "b", "a", "c", "b", "a"]

d2 = {}
for item in items:
    if item in d2:
        d2[item] += 1
    else:
        d2[item] = 1

print(d2)
```

```python
# When would you choose:
# 1. list: this is array of items support indexing like a string and be change as it is mutable
# 2. tuple: this is immutable can be used for specific type of data like a lat long data
# 3. set: this is unordered unique collection of items
# 4. dict: use key value pairs, value can be anything but keys are unique
```
